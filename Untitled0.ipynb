```python
from numpy import load
import matplotlib.pyplot as plt
import seaborn as sns

# Data load
df = sns.load_dataset("tips")

# hue='sex' addchesaim ikkada
sns.barplot(x='sex', y='tip', hue='sex', data=df, errorbar=None, palette='Set2')
plt.title('Average Tip by Gender')
plt.show()
```

```python
from matplotlib import legend
sns.barplot(x="day", y="total_bill", hue="day", data=df, legend=False)
plt.title("Total Bill by Day")
plt.show()
```

```python
sns.barplot(x='sex', y='tip', hue='sex', data=df, errorbar=None, palette='Set2')
plt.title('Average Tip by Gender')
plt.show()
```

```python
df.groupby(['day', 'time'])[['total_bill', 'tip']].mean()
```

```python
sns.pairplot(df, hue='sex', palette='husl')
plt.show()
```

```python
sns.countplot(x='day', hue='time',data=df, palette='Set1')
plt.title('Customer count by Day and Time')
plt.show()
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', style='smoker', data=df, s=70)
plt.title('Total Bill vs Tip (by Sex & Smoker)')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset first becAUSE 'df' is not defined yet
df= sns.load_dataset('tips')

# Regression plot with trend lines split by smoker status
sns.lmplot(x='total_bill', y='tip', hue='sex', col='smoker', data=df, height=4)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load the dataset first
df = sns.load_dataset('tips')

# 2. Add the new column and plot the boxplot
df['tip_percentage'] = (df['tip'] / df['total_bill']) * 100
sns.boxplot(x='day' , y='tip_percentage', hue='sex', data=df)
plt.show()
```

```python
sns.pairplot(df, hue='smoker', palette='Set2')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.select_dtypes(include=['float64', 'int64']).corr(), annot=True, cmap='coolwarm')
plt.show()
```
